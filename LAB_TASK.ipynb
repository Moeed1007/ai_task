```python
class Node:
    def __init__(self, value=None, children=None):
        self.value = value
        self.children = children if children else []

def minimax(node, is_max):
    # Base case: leaf node
    if not node.children:
        return node.value

    if is_max:
        best = float('-inf')
        for child in node.children:
            val = minimax(child, False)
            best = max(best, val)
        return best
    else:
        best = float('inf')
        for child in node.children:
            val = minimax(child, True)
            best = min(best, val)
        return best


# Leaf nodes
S = Node(-4)
T = Node(-5)
M_leaf = Node(2)
N_leaf = Node(3)
R_leaf = Node(4)
Q = Node(5)
A = Node(6)
B = Node(7)

# Level 2
D = Node(children=[S, T])         # max
R = Node(children=[M_leaf, N_leaf])  # max
O = Node(children=[R_leaf, Q])    # max
P = Node(children=[A, B])         # max

# Level 1
M = Node(children=[D, R])         # min
N = Node(children=[O, P])         # min

# Root
Z = Node(children=[M, N])         # max

# Run minimax
result = minimax(Z, True)
print("Optimal value:", result)
```

```python
class Node:
    def __init__(self, value=None, children=None):
        self.value = value
        self.children = children if children else []

def alpha_beta(node, depth, alpha, beta, is_max):
    if not node.children:
        return node.value

    if is_max:
        best = float('-inf')
        for child in node.children:
            val = alpha_beta(child, depth+1, alpha, beta, False)
            best = max(best, val)
            alpha = max(alpha, best)
            if beta <= alpha:
                break
        return best
    else:
        best = float('inf')
        for child in node.children:
            val = alpha_beta(child, depth+1, alpha, beta, True)
            best = min(best, val)
            beta = min(beta, best)
            if beta <= alpha:
                break
        return best

# Define ALL leaf nodes (based on your image's letters)
A, B, C, D = Node(3), Node(5), Node(6), Node(9)
E, F, G, H = Node(10), Node(2), Node(8), Node(4)
I, J, K, L = Node(1), Node(7), Node(12), Node(15)
M, N, O, P = Node(11), Node(14), Node(13), Node(16)

# Build tree structure (edit based on your image)
P_node = Node(children=[A, B, C, D])
Q_node = Node(children=[E, F, G, H])
R_node = Node(children=[I, J, K, L])
S_node = Node(children=[M, N, O, P])

M_node = Node(children=[P_node, Q_node])
N_node = Node(children=[R_node, S_node])

Root = Node(children=[M_node, N_node])

result = alpha_beta(Root, 0, float('-inf'), float('inf'), True)
print("Optimal value:", result)
```
